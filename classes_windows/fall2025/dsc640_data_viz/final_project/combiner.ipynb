```python
import PyPDF2

def combine_pdfs(pdf_path1, pdf_path2, output_path):
    merger = PyPDF2.PdfMerger()
    merger.append(pdf_path1)
    merger.append(pdf_path2)
    merger.write(output_path)
    merger.close()

# Example usage:
combine_pdfs('Milestone_1_Vance.pdf', 'milestone1_vance_notebook.pdf', 'Milestone_1_Vance_combined.pdf')
```

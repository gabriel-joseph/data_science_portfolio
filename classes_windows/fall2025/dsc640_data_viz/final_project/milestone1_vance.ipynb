```python
import pandas as pd

df = pd.read_excel('nationaldatabaseofchildcareprices.xlsx')
```

```python
# read the sample data
df.head()
```

```python
# check the data types and columns
print(df.dtypes)
print(df.columns)
```

```python
# print number of columns
num_columns = df.shape[1]
print(num_columns)
```

```python
# print number of rows
num_rows = df.shape[0]
print(num_rows)
```

```python
# Find columns with the same value for all rows
constant_cols = [col for col in df.columns if df[col].nunique(dropna=False) == 1]
print("Columns with the same value in all rows:", constant_cols)
print("Number of such columns:", len(constant_cols))
```

```python
# Calculate the average of the 'OneRace' column
average_one_race = df['OneRace'].mean()
print("Average of OneRace:", average_one_race)

# Create a boxplot of the 'OneRace' column
df['OneRace'].plot.box()

demographic_cols = ['OneRace_W','OneRace_B','OneRace_I','OneRace_A','OneRace_H']
```

```python
# Plot Median Earnings Female, Median Earnings, and Median Earnings Male together
df.groupby('StudyYear')[['MME_2018', 'ME_2018', 'FME_2018']].mean().plot(
    kind='line',
    title='Median Earnings by Gender and Overall',
    ylabel='Median Earnings',
    xlabel='Year'
).legend(['Median Earnings Female', 'Median Earnings', 'Median Earnings Male'])
```

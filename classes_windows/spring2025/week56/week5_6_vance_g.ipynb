Exercise 2. Create bins for each full number of earthquake magnitude (for instance, the first bin is (0,1], the second is (1,2], and so on) with the ml magnitude type and count how many are in each bin.

```python
ml_earthquakes = earthquake_df[earthquake_df["magType"] == "ml"]
max_magnitude = int(ml_earthquakes.index.max()) +1 # get max magnitude, use int to round up
print(f"Max magnitude: {max_magnitude}")
bins = range(0, max_magnitude , 1) # create bins from 0 to max_magnitude + 1 with a step of 1
ml_earthquakes['magnitude_bin'] = pd.cut(ml_earthquakes.index, bins=bins, right=True)
bin_counts = ml_earthquakes['magnitude_bin'].value_counts() # display counts of each bin

print(bin_counts)
```

Exercise 4.
Build a crosstab with the earthquake data between the tsunami column and the magType column. Rather than showing the frequency count, show the maximum magnitude that was observed for each combination. Put the magnitude type along the columns.

```python
crosstab = pd.crosstab(
    earthquake_df['tsunami'],
    earthquake_df['magType'],
    values=earthquake_df.index,
    aggfunc='max'
)

print(crosstab)
```

Calculate the z-scores for each numeric column of Amazon’s data (ticker is AMZN) in Q4 2018 using apply().

```python
faang_df.head() # print the head to see the dates, format is 2018-01-02	or YYYY-MM-DD
```

```python
amzn_df = faang_df[faang_df.index == "AMZN"]
amzn_df
# note the prices are in thousands, such as 1200, 1000
```

```python
# q4 2018 was oct1 to dec 31
q4_amzn = amzn_df[(amzn_df["date"] >= "2018-10-01") & (amzn_df["date"] <= "2018-12-31")]

# Z score formula is x - mean / std
q4_amzn_zcores = q4_amzn[["open", "high", "low", "close", "volume"]].apply(
    lambda x: (x - x.mean()) / x.std() # use apply and lambda 
)

# print q4 z scores
q4_amzn_zcores
# data is now standardized, so prices are down close to 0, but should they be negative?
```

Read in the data in the covid19_cases.csv file.

```python
covid_df = pd.read_csv("covid19_cases.csv")
covid_df.head()
```

```python
covid_df.columns # print column names
```

Create a date column by parsing the dateRep column into a datetime.

```python
covid_df['date'] = pd.to_datetime(covid_df['dateRep'], format='%d/%m/%Y')
covid_df # date is now present in the dataframe
```

Set the date column as the index.

```python
covid_df = covid_df.set_index('date') # set date as index
```

```python
covid_df.index # note that newly added date column is now the index
```

Use the replace() method to update all occurrences of United_States_of_America and United_Kingdom to USA and UK, respectively.

```python
covid_df.head()
```

```python
covid_df['countriesAndTerritories'].unique() # just to see the countries
```

```python
covid_df['countriesAndTerritories'] = covid_df['countriesAndTerritories'].replace(
    {'United_States_of_America': 'USA', 'United_Kingdom': 'UK'}
)
covid_df['countriesAndTerritories'].unique() # we can print again, note that the names changed
```

Sort the index.

```python
covid_df = covid_df.sort_index(ascending=True) # sort but have to make it ascending to earlier dates are first
covid_df
```

For the five countries with the most cases (cumulative), find the day with the largest number of cases.
Find the 7-day average change in COVID-19 cases for the last week in the data for the five countries with the most cases.

```python
total_cases = covid_df.groupby('countriesAndTerritories')['cases'].sum() # this gets the sum of all cases by country
top_countries = total_cases.sort_values(ascending=False).head(5).index # top 5
country_names = top_countries.to_list()
# top countries are USA, India, Brazil, Russia, Peru
country_names
```

```python
def parse_country_data(country):
    country_df = covid_df[covid_df['countriesAndTerritories'] == country] # filter for country
    sorted_by_cases = country_df.sort_values(by='cases',ascending=True) # sort by cases, ascending
    top_date = sorted_by_cases.iloc[-1] # get the date of the last case
    top_date_value = top_date.name
    # average change - the mean difference between 7 consecutive days
    last_week = country_df.tail(7)
    last_week['average_change'] = last_week['cases'].diff()
    average = last_week['average_change'].mean()
    print(f"{country}'s top date for COVID-19 cases is {top_date_value}. The average change in the last week is {average:.2f} cases.")

for country in country_names:
    parse_country_data(country) 
```

Find the first date that each country other than China had cases.


```python
all_countries = total_cases.sort_values(ascending=False).index.to_list()
"China" in all_countries
```

```python
all_countries = total_cases.sort_values(ascending=False).index.to_list()
all_countries.remove("China")
"China" in all_countries # check if China is removed
```

```python
all_countries
```

```python
def determine_first_date_of_covid_cases(country):
    country_df = covid_df[(covid_df['countriesAndTerritories'] == country) & (covid_df['cases'] > 0)] # filter for country AND cases greater than 0
    sorted_by_cases = country_df.sort_values(by='cases',ascending=True) # sort by cases, ascending
    first_date = sorted_by_cases.iloc[0] # get the date of the first case
    first_date_value = first_date.name
    print(f"{country}'s first date for COVID-19 cases is {first_date_value}.")
    
for country in all_countries:
    determine_first_date_of_covid_cases(country) 
```

Rank the countries by cumulative cases using percentiles.

## Read CSVs into dfs and save results to a single CSV

```python
import pandas as pd
```

### Dataframe concation

```python
# Read all CSVs into dataframes
aapl_df = pd.read_csv("aapl.csv")
amzn_df = pd.read_csv("amzn.csv")
fb_df = pd.read_csv("fb.csv")
goog_df = pd.read_csv("goog.csv")
nflx_df = pd.read_csv("nflx.csv")

# Add ticker column
aapl_df['ticker'] = "AAPL"
amzn_df['ticker'] = "AMZN"
fb_df['ticker'] = "FB"
goog_df['ticker'] = "GOOG"
nflx_df['ticker'] = "NFLX"
```

## Exercise 4 
## Change wide data to long using melt

### Note: 
### Wide data means a single row with multiple time points. 
### Long data would break out the columns into new rows. 

Wide Example: 

Store, Jan Sales, Feb Sales, Mar Sales

Long Example:
Store, Sales, Month (Jan, Feb, Mar)

melt() will allow us to convert from wide to long

```python
# Use melt() to make it completely long format.

# id_vars (not melted). Think keys/composite keys.
id_vars = ['ticker','date']
value_vars = ["open", "high", "low", "close", "volume"]
# var_name is the name of the new column to represent data types (think categorical for stocks)
# value_name will contain the actual data from the original columns
faang_long = faang_df.melt(id_vars=id_vars, value_vars=value_vars,var_name='long metric', value_name='long value')
print(faang_long)
```

## Exercise 6

* Read Covid data into dataframe
* Create column using the dateRep column and convert to datetime
* Set date column to index
* Replace long name with short name for USA/UK (use replace())
* Filter data using countries column for Argentina, Brazil, China, Colombia, India, Italy, Mexico, Peru, Russia, Spain, Turkey, the UK, and the USA.
* Fill NaN with 0. Pivot so date is index, Columns contain country names, values are cases

```python
# read to dataframe
covid19_data = pd.read_csv("covid19.csv")



# create column
covid19_data['DateRep_as_DateTime'] = pd.to_datetime(covid19_data['dateRep'], format='%d/%m/%Y')
# set as index
covid19_data.set_index('DateRep_as_DateTime', inplace=True)

# short name/long name dict 
short_name_long_name = {
    'United States': 'USA',
    'United Kingdom': 'UK'
}

# replace long names with short names
covid19_data['countriesAndTerritories'] = covid19_data['countriesAndTerritories'].replace(short_name_long_name)

# Make a list of countries we want to use/keep
countries_list = [
    'Argentina', 'Brazil', 'China', 'Colombia', 'India', 'Italy', 
    'Mexico', 'Peru', 'Russia', 'Spain', 'Turkey', 'United Kingdom', 'USA'
]

# filter df if country in country lsit
filtered_data = covid19_data[covid19_data['countriesAndTerritories'].isin(countries_list)]

# double check the state/shape of filtered data, note that Argentina is now the first country instead of afghanistan
filtered_data
```

```python
# fill NAs to be 0
filtered_data.fillna(0, inplace=True)

# Pivot the dataframe
pivoted_data = filtered_data.pivot_table(
    index='DateRep_as_DateTime', 
    columns='countriesAndTerritories', 
    # using cases to be the values for the rows
    values='cases', 
    # the sum function wasn't necessary as rows didn't have the same date, but this got rid of floats
    aggfunc='sum', 
    fill_value=0 # if there are any leftover values
)
# pivot table with cases as values
pivoted_data
```

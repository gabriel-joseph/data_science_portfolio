```python
from pathlib import Path
from PyPDF2 import PdfMerger
from pypdf import PdfMerger



src_dir = Path("/Users/gvance67/GitHub/data_science_classes_windows/fall2025/dsc640_data_viz/final_project/milestone3/milestone_pdf")
output_file = src_dir / "merged_milestone3.pdf"

pdf_files = sorted(src_dir.glob("*.pdf"))
if not pdf_files:
    raise FileNotFoundError(f"No PDF files found in {src_dir}")

merger = PdfMerger()
for pdf in pdf_files:
    merger.append(str(pdf))

with open(output_file, "wb") as f:
    merger.write(f)
merger.close()

print(f"Merged {len(pdf_files)} PDFs into: {output_file}")
```

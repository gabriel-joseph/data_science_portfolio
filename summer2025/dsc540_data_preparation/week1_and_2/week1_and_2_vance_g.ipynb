### Create a list of 100 random numbers. 


```python
import random

random_numbers = [random.randint(1, 1000) for _ in range(100)]
print(random_numbers)
```

### Create a new list from this random list, with numbers that are divisible by 3. 

```python
divisible_by_3 = [num for num in random_numbers if num % 3 == 0]
print(divisible_by_3)
```

### Calculate the length of these two lists and store the difference in a new variable.

```python
len_random_numbers = len(random_numbers)
len_divisible_by_3 = len(divisible_by_3)
length_difference = len_random_numbers - len_divisible_by_3
print(length_difference)
```

### Using a loop, perform steps 1, 2, and 3, and find the difference variable 10 times.

```python
differences = []
for _ in range(10):
    nums = [random.randint(1, 1000) for _ in range(100)]
    div_by_3 = [num for num in nums if num % 3 == 0]
    diff = len(nums) - len(div_by_3)
    differences.append(diff)
print(differences)
```

The goal of this activity is to fetch data and retrieve information from two tables, persons and pets, which are a part of the petsdb database.

```python
# connect to db
import sqlite3
conn = sqlite3.connect("petsdb")  

# check to see if successful 

def is_opened(conn):
    try:
        conn.execute("SELECT * FROM persons LIMIT 1")
        return True
    except sqlite3.ProgrammingError as e:
        print("Connection closed {}".format(e))
        return False
print(is_opened(conn))
```

```python
# We can also close and open the connection

conn.close()
print(is_opened(conn))
```

```python
# Open connection and create cursor
conn = sqlite3.connect("petsdb")
c = conn.cursor()

# What is the count of people belonging to different age groups in the persons table?
for ppl, age in c.execute("SELECT count(*), age FROM persons GROUP BY age"):
    print("We have {} people aged {}".format(ppl, age))
```

```python
# Which age group has the maximum number of people?
for ppl, age in c.execute(
    "SELECT count(*), age FROM persons GROUP BY age ORDER BY count(*) DESC"):
    print("The highest number of people is {} and came from {} age group".format(ppl, age))
    break
```

```python
# How many people do not have a last name?
for ppl in c.execute("SELECT count(*) FROM persons WHERE last_name IS NULL"):
    print("There are {} people without a last name.".format(ppl))
```

```python
# How many people have more than one pet?
for ppl in c.execute("SELECT count(*) FROM persons WHERE id IN (SELECT owner_id FROM pets GROUP BY owner_id HAVING count(*) > 1)"):
    print("There are {} people who have more than one pet.".format(ppl))
```

```python
# How many pets received treatment?
res = c.execute("SELECT count(*) FROM pets WHERE treatment_done=1")
for row in res:
    print(row)
```

```python
# How many pets have received treatment, and the type of pet is known?
res = c.execute("SELECT count(*) FROM pets WHERE treatment_done=1 AND pet_type IS NOT null")
for row in res:
    print(row)
```

```python
# How many pets are from the city called east port?
res = c.execute("SELECT count(*) FROM pets JOIN persons ON pets.owner_id = persons.id WHERE persons.city='east port'")
for row in res:
    print(row)
```

```python
# How many pets are from the city called east port, and who received treatment?
res = c.execute("SELECT count(*) FROM pets JOIN persons ON pets.owner_id = persons.id WHERE persons.city='east port' AND pets.treatment_done=1")
for row in res:
    print(row)
```

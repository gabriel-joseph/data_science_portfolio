## Find the top five games by critic score.


```python
# Sort df by Critic_Score
top5_critic = df.sort_values(by='Critic_Score', ascending=False)
# Print the top 5 rows of the sorted DataFrame
top5_critic.head(5)
```

## Find the number of video games in the data frame in each genre.


```python
# Use value_counts() to count the number of games in each genre
genre_counts = df['Genre'].value_counts()
genre_counts
```

## Find the first five games in the data frame on the SNES platform.


```python
# Display the first 5 rows, using the sorted df. Filter by Platform 'SNES'
top5_critic[top5_critic['Platform'] == 'SNES'].head(5)
```

## Find the five publishers with the highest total global sales. Note: You will need to calculate the total global sales for each publisher to do this.


```python
# Group the data by 'Publisher'
grouped = df.groupby('Publisher')

# Sum the 'Global_Sales'
publisher_sales = grouped['Global_Sales'].sum()

# Sort the summed sales in descending order. This gives us the highest total sales.
sorted_sales = publisher_sales.sort_values(ascending=False)

# Select the top 5 publishers
top5_publishers = sorted_sales.head(5)

top5_publishers
```

## Create a new column in the data frame that calculates the percentage of global sales from North America. Display the first five rows of the new data frame.


```python
# Divide NA sales by Global Sales to get the percentage of NA sales. We could also add all of the other regions and the NA together.
df['NA_Global_Sales_As_A_Percentage'] = (df['NA_Sales'] / df['Global_Sales']) * 100
df.head(5)
```

## Find the number NaN entries (missing data values) in each column.


```python
# Get the NA items and their sum
missing_values = df.isna().sum()
# Print the missing values for each column
missing_values
```

## Try to calculate the median user score of all the video games. You will likely run into an error because some of the user score entries are a non-numerical string that cannot be converted to a float. Find and replace this string with NaN and then calculate the median. Then, replace all NaN entries in the user score column with the median value.


```python
# Convert 'User_Score' to numeric, coercing errors to NaN
df['User_Score'] = pd.to_numeric(df['User_Score'], errors='coerce')

# Calculate the median, ignoring NaN values
median_user_score = df['User_Score'].median()

# Replace NaN values with the median
df['User_Score'].fillna(median_user_score, inplace=True)

print(f"Median User Score: {median_user_score}")
df.head()
```

```python
import pandas as pd
import numpy as np
```

### Import the housing data as a data frame and ensure that the data is loaded properly.

```python
test_df = pd.read_csv('test.csv')
test_df.head()
```

### Drop the "Id" column and any features that are missing more than 40% of their values.

```python
# Drop the "Id" column
test_df = test_df.drop(columns=['Id'])

# Drop columns with more than 40% missing values
threshold = len(test_df) * 0.6
test_df = test_df.dropna(axis=1, thresh=threshold)

test_df.head()
```

### For numerical columns, fill in any missing data with the median value.

```python
# Fill missing values in numerical columns with the median
num_cols = test_df.select_dtypes(include=[np.number]).columns
test_df[num_cols] = test_df[num_cols].fillna(test_df[num_cols].median())
test_df.head()
```

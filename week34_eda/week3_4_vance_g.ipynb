Exercise 7.
Scale the data created in Exercise 4 using the following strategies:

```python
# min max scaling formula: (X - X_min) / (X_max - X_min)

X_min = np.min(salaries)
X_max = np.max(salaries)
normalized_data = (salaries - X_min) / (X_max - X_min)
normalized_data


mean = np.mean(salaries)
std_dev = np.std(salaries)

# Standardization
standardized_data = (salaries - mean) / std_dev

standardized_data
```

Exercise 8.
Using the scaled data from Exercise 7, calculate the following:

```python
# Assuming normalized_data and standardized_data are NumPy arrays of the same length
cov_matrix = np.cov(normalized_data, standardized_data)

# Extract covariance value (off-diagonal element)
covariance = cov_matrix[0, 1]

print("Covariance Matrix:\n", cov_matrix)
print("Covariance between normalized and standardized data:", covariance)




# Assuming normalized_data and standardized_data are NumPy arrays of the same length
correlation_matrix = np.corrcoef(normalized_data, standardized_data)

# Extract correlation value
pearson_corr = correlation_matrix[0, 1]

print("Correlation Matrix:\n", correlation_matrix)
print("Pearson Correlation Coefficient:", pearson_corr)
```
